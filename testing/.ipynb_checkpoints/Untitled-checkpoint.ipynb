```python
import requests
import json

ADDRESS = "c9e16de7648231d030be33c053b11ae721121b8391db7af97bf0fa779396728a"

url = f"https://api.blockchair.com/bitcoin/dashboards/address/{ADDRESS}"

response = requests.get(url, timeout=30)

print("HTTP Status:", response.status_code)

if response.status_code != 200:
    print(response.text)
    exit()

data = response.json()

address_data = data["data"][ADDRESS]

print("\n==============================")
print("BITCOIN ADDRESS")
print("==============================")

print("Address :", ADDRESS)
print("Balance :", address_data.get("address", {}).get("balance"))
print("Received:", address_data.get("address", {}).get("received"))
print("Spent   :", address_data.get("address", {}).get("spent"))

print("\nTransactions:")

for tx in address_data.get("transactions", []):
    print(tx)
```
